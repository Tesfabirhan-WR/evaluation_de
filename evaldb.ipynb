Python pour analyse des Ventes de la PME

```python
import sqlite3
import pandas as pd
```

Changement des données Excel

```python
df_sales = pd.read_excel(r"\evaluationDE\data.sqlite\sales.xlsx")
#df_sales.head()
```

```python
df_stores = pd.read_excel(r"\evaluationDE\data.sqlite\stores.xlsx")
#df_stores.head()
```

```python
df_products = pd.read_excel(r"\evaluationDE\data.sqlite\products.xlsx")
#df_products.head()
```

Connect sqlite3

```python
conn = sqlite3.connect("sales.db")
```

```python
df_sales.to_sql("sales", conn, if_exists="replace", index=False)
df_products.to_sql("products", conn, if_exists="replace", index=False)
df_stores.to_sql("stores", conn, if_exists="replace", index=False)

conn.commit()
conn.close()

print("Données insérées avec succès dans SQLite.")
```
